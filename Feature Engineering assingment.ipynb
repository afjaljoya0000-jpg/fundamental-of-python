1. What is a parameter?

A parameter is a numerical value inside a model that is learned from the data during training. For example, the slope and intercept in linear regression are parameters.

2. What is correlation?

Correlation is a statistical relationship that shows how two variables move with respect to each other.

What does negative correlation mean?

Negative correlation means that when one variable increases, the other variable decreases. They move in opposite directions.

3. Define Machine Learning. What are the main components in Machine Learning?

Machine Learning is a field of computer science where algorithms learn patterns from data without being explicitly programmed.

Main components:

Data

Model

Loss function

Optimizer/Training algorithm

Evaluation metrics

4. How does loss value help in determining whether the model is good or not?

Loss value measures the error between predicted and actual values.

Low loss → good model

High loss → poor model
The goal of training is to minimize the loss.

5. What are continuous and categorical variables?

Continuous variables are numerical and can take any value (e.g., height, salary).

Categorical variables represent categories or labels (e.g., gender, color, city).

6. How do we handle categorical variables in Machine Learning? What are the common techniques?

We convert categories into numbers. Common techniques:

Label Encoding

One-Hot Encoding

Ordinal Encoding

Frequency Encoding

7. What do you mean by training and testing a dataset?

Training dataset: used to teach the model.

Testing dataset: used to evaluate how well the model performs on unseen data.

8. What is sklearn.preprocessing?

sklearn.preprocessing is a module in scikit-learn that provides functions for data preprocessing such as encoding, scaling, normalization, and transformation.

9. What is a Test set?

A test set is a portion of data used only to evaluate the final performance of a model after training.

16. What is causation? Explain difference between correlation and causation with an example.

Causation means one variable directly affects another.

Difference:

Correlation = relationship

Causation = cause-and-effect

Example:
Ice cream sales and drowning cases are correlated, but ice cream does not cause drowning. The cause is summer/heat.

17. What is an Optimizer? What are different types of optimizers? Explain with an example.

An optimizer is an algorithm that updates model parameters to minimize the loss function.

Types:

Gradient Descent

SGD

Adam

RMSProp

Example: Adam optimizer adjusts learning rates automatically to improve training speed.

Feature scaling transforms features into a similar scale.
