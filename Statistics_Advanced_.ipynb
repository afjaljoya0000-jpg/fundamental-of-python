#Question 7: What is the significance of confidence intervals in statistical analysis?

#ANSWER-Significance of Confidence Intervals
1. Shows the Precision of an Estimate

A narrow CI means the estimate is very precise.
A wide CI means the estimate is less reliable.

2. Gives More Information Than a Single Number

Instead of giving only a sample mean (e.g., 50), a CI provides a range (e.g., 48 to 52), which is more informative.

3. Helps in Decision-Making

Confidence intervals help decide if a value (like a target or claim) is reasonable.
Example: Checking if a new medicine improves recovery time.

4. Basis for Hypothesis Testing

If the CI does not include the hypothesized value (e.g., μ = 0), it suggests significant evidence against the null hypothesis.

5. Quantifies Uncertainty

CI provides a clear way to understand how much uncertainty exists in sample-based estimates of population parameters.

#Question 9: Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
deviation using NumPy, and draw a histogram to visualize the distribution.
(Include your Python code and output in the code box below.)

#ANSWER-Here is a complete Python program that:

✔ Generates 1000 random numbers from a normal distribution
✔ Computes mean and standard deviation using NumPy
✔ Draws a histogram using Matplotlib
✔ Includes sample output

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random numbers from N(50, 5)
data = np.random.normal(loc=50, scale=5, size=1000)

# Compute mean and standard deviation
mean_value = np.mean(data)
std_value = np.std(data)

print("Calculated Mean:", mean_value)
print("Calculated Standard Deviation:", std_value)

# Plot histogram
plt.hist(data, bins=30, edgecolor='black')
plt.title("Histogram of Normal Distribution (Mean=50, SD=5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

Question 10: You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
 235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
● Explain how you would apply the Central Limit Theorem to estimate the average sales
with a 95% confidence interval.
● Write the Python code to compute the mean sales and its confidence interval.
(Include your Python code and output in the code box below.)#

#ANSWER-The company has daily sales data. To estimate the true average daily sales, we use the sample mean and sample standard deviation.

The Central Limit Theorem (CLT) says:

When we take repeated samples from any population, the sampling distribution of the sample mean becomes approximately normal, even if the data is not perfectly normal — especially when sample size ≥ 30.

Here, we only have 20 values, but CLT still allows us to use the t-distribution for confidence intervals when sample size is small.

```python
import numpy as np
from scipy import stats

# Daily sales data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to NumPy array
sales = np.array(daily_sales)

# Calculate mean and standard deviation
mean_sales = np.mean(sales)
std_sales = np.std(sales, ddof=1)   # sample standard deviation

# Sample size
n = len(sales)

# Standard error of the mean (SEM)
sem = std_sales / np.sqrt(n)

# t-critical value for 95% CI with df = n - 1
t_critical = stats.t.ppf(0.975, df=n-1)

# Confidence interval
lower_bound = mean_sales - t_critical * sem
upper_bound = mean_sales + t_critical * sem

print("Mean Sales:", mean_sales)
print("Standard Deviation:", std_sales)
print("95% Confidence Interval: ({:.2f}, {:.2f})".format(lower_bound, upper_bound))
```
