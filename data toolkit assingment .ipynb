1. What is NumPy, and why is it widely used in Python?

NumPy (Numerical Python) is a fundamental library used for numerical computing in Python. It provides support for large multi-dimensional arrays and matrices along with a collection of mathematical functions to operate on them efficiently. It is widely used because it is faster than Python lists, memory-efficient, and forms the base for many scientific and data analysis libraries.

2. How does broadcasting work in NumPy?

Broadcasting allows NumPy to perform arithmetic operations on arrays of different shapes. Instead of copying data, NumPy stretches smaller arrays to match the shape of larger ones so that operations can be performed element-wise efficiently.

3. What is a Pandas DataFrame?

A Pandas DataFrame is a two-dimensional labeled data structure similar to a table or spreadsheet. It consists of rows and columns where each column can contain different data types, making it ideal for data analysis and manipulation.

4. Explain the use of the groupby() method in Pandas.

The groupby() method is used to split data into groups based on one or more columns, apply aggregation functions like sum, mean, or count, and then combine the results. It is useful for analyzing data based on categories.

5. Why is Seaborn preferred for statistical visualizations?

Seaborn is preferred because it provides high-level statistical plotting functions, attractive default styles, and easy integration with Pandas DataFrames. It simplifies the creation of complex statistical plots such as box plots, violin plots, and regression plots.

6. What are the differences between NumPy arrays and Python lists?

NumPy arrays are faster, consume less memory, and support vectorized operations, whereas Python lists are more flexible but slower for numerical computations. NumPy arrays store elements of the same data type, while lists can store mixed data types.

7. What is a heatmap, and when should it be used?

A heatmap is a graphical representation of data where values are represented by colors. It is commonly used to visualize correlation matrices, patterns, and intensity of data across two dimensions.

8. What does the term “vectorized operation” mean in NumPy?

Vectorized operations refer to performing operations on entire arrays at once without using explicit loops. This improves performance and makes the code more readable and efficient.

9. How does Matplotlib differ from Plotly?

Matplotlib is a static plotting library mainly used for basic and publication-quality graphs. Plotly is an interactive visualization library that supports zooming, hovering, and web-based dashboards.

10. What is the significance of hierarchical indexing in Pandas?

Hierarchical indexing, also known as MultiIndexing, allows multiple index levels on rows or columns. It helps represent higher-dimensional data in a two-dimensional structure and simplifies complex data analysis.

11. What is the role of Seaborn’s pairplot() function?

The pairplot() function is used to visualize pairwise relationships between numerical variables in a dataset. It helps in understanding correlations, distributions, and trends in data.

12. What is the purpose of the describe() function in Pandas?

The describe() function provides a quick statistical summary of numerical data, including count, mean, standard deviation, minimum, maximum, and quartiles.

13. Why is handling missing data important in Pandas?

Handling missing data is important to maintain data accuracy and reliability. Missing values can lead to incorrect analysis and results, so Pandas provides methods like fillna() and dropna() to manage them.

14. What are the benefits of using Plotly for data visualization?

Plotly offers interactive visualizations, supports 3D plots, integrates well with web applications, and allows users to explore data dynamically through zooming and hovering.

15. How does NumPy handle multidimensional arrays?

NumPy uses the ndarray object to handle multidimensional arrays. These arrays can have any number of dimensions and allow efficient mathematical operations across multiple axes.

16. What is the role of Bokeh in data visualization?

Bokeh is used to create interactive and scalable visualizations for modern web browsers. It is especially useful for building dashboards and real-time data applications.

17. Explain the difference between apply() and map() in Pandas.

The apply() function works on rows or columns of a DataFrame, while map() works on individual elements of a Series. apply() is more flexible and powerful.

18. What are some advanced features of NumPy?

Advanced features of NumPy include broadcasting, vectorization, linear algebra operations, random number generation, Fourier transforms, and memory-efficient array handling.

19. How does Pandas simplify time series analysis?

Pandas provides date-time indexing, resampling, shifting, rolling window operations, and easy time-based slicing, making time series analysis simple and efficient.

20. What is the role of a pivot table in Pandas?

A pivot table summarizes and reshapes data by aggregating values based on categories. It helps in analyzing large datasets by creating meaningful summaries.

21. Why is NumPy’s array slicing faster than Python’s list slicing?

NumPy slicing returns a view instead of a copy and operates at the C-level, making it faster and more memory-efficient than Python list slicing.

22. What are some common use cases for Seaborn?

Common use cases include exploratory data analysis, statistical comparison of datasets, visualizing distributions, correlations, and regression relationships.

1. How do you create a 2D NumPy array and calculate the sum of each row?

```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

row_sum = arr.sum(axis=1)
print(row_sum)
```

2. Write a Pandas script to find the mean of a specific column in a DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    'Marks': [70, 80, 90, 85]
})

mean_value = df['Marks'].mean()
print(mean_value)
```

3. Create a scatter plot using Matplotlib.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.scatter(x, y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot")
plt.show()
```

4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'A': [10, 20, 30],
    'B': [20, 30, 40],
    'C': [30, 40, 50]
})

corr = df.corr()
sns.heatmap(corr, annot=True)
plt.show()
```

5. Generate a bar plot using Plotly.

```python
import plotly.express as px

data = {'Category': ['A', 'B', 'C'], 'Values': [10, 20, 30]}
fig = px.bar(data, x='Category', y='Values', title='Bar Plot')
fig.show()
```

6. Create a DataFrame and add a new column based on an existing column.

```python
import pandas as pd

df = pd.DataFrame({'Salary': [20000, 30000, 40000]})
df['Bonus'] = df['Salary'] * 0.10
print(df)
```

7. Write a program to perform element-wise multiplication of two NumPy arrays.

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = a * b
print(result)
```

8. Create a line plot with multiple lines using Matplotlib.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [10, 20, 30, 40]
y2 = [15, 25, 35, 45]

plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')
plt.legend()
plt.show()
```

9. Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.

```python
import pandas as pd

df = pd.DataFrame({'Age': [18, 25, 30, 16]})
filtered_df = df[df['Age'] > 20]
print(filtered_df)
```

11. Perform matrix multiplication using NumPy.

```python
import numpy as np

a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])

result = np.dot(a, b)
print(result)
```
